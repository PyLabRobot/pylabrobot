# Thermo Fisher NanoDrop 1000

The NanoDrop 1000 measures absorbance spectra from small-volume samples. Complete the [NanoDrop 1000 setup guide](setup.md) before connecting with PyLabRobot.

## Connect

```python
from pylabrobot.thermo_fisher.nanodrop_1000 import ThermoFisherNanoDrop1000

nanodrop = ThermoFisherNanoDrop1000()
await nanodrop.setup()
```

## Blank

Clean the pedestals, load the blank solution, lower the sampling arm, and acquire the blank.

```python
# Blanks both path lengths, so either can be used without re-blanking.
await nanodrop.take_blank(integration_ms=20)
```

## Measure a sample

Clean the pedestals, load the sample, lower the sampling arm, and measure absorbance. The result contains matching wavelength and absorbance lists.

```python
# A normal run measures BOTH paths, as the instrument does.
wavelengths, spectra = await nanodrop.measure_absorbance(integration_ms=20)

for path_mm, absorbance in sorted(spectra.items(), reverse=True):
    peak = max(range(len(absorbance)), key=absorbance.__getitem__)
    print(f"{path_mm} mm: {absorbance[peak]:.3f} A at {wavelengths[peak]:.1f} nm")

# Quantify from whichever path is in range, quoted against a 10 mm cuvette.
best = nanodrop.select_path(wavelengths, spectra)
equivalent = ThermoFisherNanoDrop1000.to_path_length(spectra[best], from_mm=best)
peak = max(range(len(equivalent)), key=equivalent.__getitem__)
print(f"Using the {best} mm path: {equivalent[peak]:.2f} A at 10 mm equivalent")
```

## Disconnect

```python
await nanodrop.stop()
```
