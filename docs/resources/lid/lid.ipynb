# Lid

A {class}`~pylabrobot.resources.Lid` is a removable cover for a plate, a tip rack or a container.
It is a resource of its own: a lid on a plate is one of the plate's children, next to its wells,
and it can be taken off and put back.

## Creating a lid

Besides a name and its outer dimensions, a lid has a `nesting_z_height`: the overlap, in mm,
between the lid and the resource it covers, in the z direction. `size_z` is the lid's full height.
Both can be measured with a caliper:

![Measuring size_z and nesting_z_height](../img/plate/lid_nesting_z_height.jpeg)

The lid below has the dimensions of `cor_96_wellplate_360uL_Fb_lid` in the resource library, the
lid of a Corning 96-well plate:

```python
from pylabrobot.resources import Lid

lid = Lid(
  name="plate_lid",
  size_x=127.76,
  size_y=85.48,
  size_z=8.9,
  nesting_z_height=7.6,
)
```

## Putting a lid on

Set the plate's `lid`, or assign the lid to the plate with `assign_child_resource`. Unless a
location is given, the lid is centred on the plate's top face and lowered by its
`nesting_z_height`. The lid's location, like any child's, is relative to the plate:

```python
from pylabrobot.resources import Coordinate, cor_96_wellplate_360uL_Fb

plate = cor_96_wellplate_360uL_Fb(name="plate")  # 14.2 mm tall
plate.lid = lid  # the same as plate.assign_child_resource(lid)

assert plate.lid is lid
assert lid.location == Coordinate(0, 0, 14.2 - 7.6)
```

`Plate(..., lid=lid)` puts a lid on when the plate is created. Some library plates
do this when called with `with_lid=True`: `cor_96_wellplate_360uL_Fb(name="plate",
with_lid=True)` creates the same plate with the library's `cor_96_wellplate_360uL_Fb_lid` on,
named `plate_lid`.

The lid is one of the plate's `children`, next to its wells. The plate's item methods, such as
`get_all_items()`, return only the wells.

## Taking a lid off

Unassign the lid, or set the plate's `lid` to `None`. The lid then has no parent and no location:

```python
lid.unassign()  # the same as plate.lid = None

assert not plate.has_lid()
assert lid.parent is None and lid.location is None
```

Putting a lid on and taking it off change only the resource model; they do not move a physical
lid.

## Resources that take a lid

A resource takes a lid through the `Liddable` mixin, which provides `lid`, `has_lid()` and
`get_lid_location()`, the location used above. {class}`~pylabrobot.resources.Plate`,
{class}`~pylabrobot.resources.TipRack` and {class}`~pylabrobot.resources.Container` (troughs,
tubes, petri dishes and wells) are liddable; a lid is not.

A liddable resource holds one lid at a time: putting on a second raises `ValueError`. So does a lid
more than 1 mm smaller than the resource in x or y, which could not cover it. A larger lid is
centred, so it overhangs evenly:

```python
from pylabrobot.resources import hamilton_1_trough_200mL_Vb

trough = hamilton_1_trough_200mL_Vb(name="trough")  # 37 x 118 x 95 mm
trough_lid = Lid(
  name="trough_lid",
  size_x=39,
  size_y=120,
  size_z=10,
  nesting_z_height=4,
)
trough.lid = trough_lid

assert trough_lid.location == Coordinate(-1, -1, 95 - 4)
```

See {class}`~pylabrobot.resources.Lid` for the API reference.
